```python
"""
calibrate_v2.py — 修正版标定
"""
import RocketSim as rs
import numpy as np
import json

# ---- 正确的方向映射 ----
DIR_MAP = {
    "front":       (1, 0),
    "back":        (-1, 0),
    "left":        (0, -1),
    "right":       (0, 1),
    "front_left":  (1, -1),
    "front_right": (1, 1),
    "back_left":   (-1, -1),
    "back_right":  (-1, 1),
}

def make():
    arena = rs.Arena(rs.GameMode.SOCCAR)
    car = arena.add_car(rs.Team.BLUE, rs.CarConfig.OCTANE)
    return arena, car

def place(car, x=3000, y=3000, z=17, vel=None):
    cs = rs.CarState()
    cs.pos = rs.Vec(x, y, z)
    if vel is not None:
        cs.vel = rs.Vec(*vel)
    car.set_state(cs)

def zero():
    return rs.CarControls()

def snap(car):
    cs = car.get_state()
    fwd = car.get_forward_dir()
    return {
        'pos': [cs.pos.x, cs.pos.y, cs.pos.z],
        'vel': [cs.vel.x, cs.vel.y, cs.vel.z],
        'pitch_deg': float(np.degrees(np.arctan2(fwd.z, np.hypot(fwd.x, fwd.y)))),
        'fwd': [fwd.x, fwd.y, fwd.z],
        'on_ground': bool(cs.is_on_ground),
        'is_flipping': bool(cs.is_flipping),
        'is_jumping': bool(cs.is_jumping),
        'has_flip_or_jump': bool(cs.has_flip_or_jump),
        'air_time': float(cs.air_time),
        'air_time_since_jump': float(cs.air_time_since_jump),
    }

# ============================================================
# 标定 1: jump hold 上限（测到 60 帧）
# ============================================================
def calib_jump_hold_v2():
    out = {}
    for hold in [0,1,2,3,4,5,6,8,10,12,14,16,18,20,25,30,40,60]:
        arena, car = make()
        place(car)
        arena.step(1)
        
        if hold > 0:
            c = zero(); c.jump = True
            car.set_controls(c)
            for _ in range(hold):
                arena.step(1)
        car.set_controls(zero())
        
        max_z = 0.0
        landed_at = None
        for t in range(180):
            arena.step(1)
            s = snap(car)
            max_z = max(max_z, s['pos'][2])
            if s['on_ground'] and t > hold and landed_at is None:
                landed_at = t
        
        out[hold] = {
            'max_z': max_z,
            'air_time_frames': landed_at,
        }
    return out

# ============================================================
# 标定 2: dodge 速度加成 vs 初始速度
# ============================================================
def calib_dodge_speed_boost():
    """在不同初始速度下 dodge，测 dodge 结束时的水平速度"""
    out = {}
    for init_vx in [0, 500, 1000, 1500, 2000, 2200]:
        arena, car = make()
        place(car, vel=(init_vx, 0, 0))
        arena.step(1)
        
        # jump
        c = zero(); c.jump = True
        car.set_controls(c); arena.step(1)
        car.set_controls(zero())
        
        # 等 5 帧
        for _ in range(5):
            arena.step(1)
        
        # dodge（前翻）
        thr, st = DIR_MAP["front"]
        c = zero(); c.jump = True; c.throttle = thr; c.steer = st
        car.set_controls(c); arena.step(1)
        car.set_controls(zero())
        
        # 观察 dodge 结束后速度
        vx_at_end = 0.0
        for _ in range(60):
            arena.step(1)
            s = snap(car)
            if not s['is_flipping'] and s['on_ground']:
                vx_at_end = abs(s['vel'][0])
                break
        
        out[init_vx] = {
            'vx_after_dodge': vx_at_end,
            'gain': vx_at_end - init_vx,
        }
    return out

# ============================================================
# 标定 3: jump hold × dodge 刮地
# ============================================================
def calib_dodge_height():
    """不同 hold 帧数起跳后 dodge，看是否刮地"""
    out = {}
    for hold in [1, 3, 5, 8, 12, 16, 20]:
        arena, car = make()
        place(car)
        arena.step(1)
        
        # jump
        c = zero(); c.jump = True
        car.set_controls(c)
        for _ in range(hold):
            arena.step(1)
        car.set_controls(zero())
        
        # 等 5 帧
        for _ in range(5):
            arena.step(1)
        
        # dodge
        thr, st = DIR_MAP["front"]
        c = zero(); c.jump = True; c.throttle = thr; c.steer = st
        car.set_controls(c); arena.step(1)
        car.set_controls(zero())
        
        min_z = 1e9
        for _ in range(60):
            arena.step(1)
            s = snap(car)
            min_z = min(min_z, s['pos'][2])
        
        out[hold] = {
            'min_z_after_dodge': min_z,
            'scraped': min_z < 17.5,
        }
    return out

# ============================================================
# 标定 4: flip cancel 逐帧曲线
# ============================================================
def calib_flip_cancel_v2():
    """
    前翻（throttle=1）→ 第 N 帧开始按 S（throttle=-1）→ 记录 pitch 曲线
    同时测：
      A) 按住 S 直到 dodge 结束
      B) 按住 S 到一半就松开（测 cancel cancel）
    """
    out = {}
    for cancel_at in [0, 2, 4, 6, 8, 10, 12, 14, 16]:
        # --- A: 按住 S 到底 ---
        arena, car = make()
        place(car)
        arena.step(1)
        c = zero(); c.jump = True
        car.set_controls(c); arena.step(1)
        car.set_controls(zero())
        for _ in range(5): arena.step(1)
        
        # 触发前翻 (W)
        c = zero(); c.jump = True; c.throttle = 1
        car.set_controls(c); arena.step(1)
        car.set_controls(zero())
        
        trace_A = []
        # 等 cancel_at 帧
        for _ in range(cancel_at):
            arena.step(1)
        
        # 按住 S 到 dodge 结束
        for t in range(60):
            c = zero(); c.throttle = -1
            car.set_controls(c)
            arena.step(1)
            s = snap(car)
            trace_A.append({'t': t, 'pitch': s['pitch_deg'], 'flipping': s['is_flipping']})
            if not s['is_flipping'] and t > 3:
                break
        
        # --- B: 按住 S 一段时间后松开 ---
        arena2, car2 = make()
        place(car2)
        arena2.step(1)
        c = zero(); c.jump = True
        car2.set_controls(c); arena2.step(1)
        car2.set_controls(zero())
        for _ in range(5): arena2.step(1)
        c = zero(); c.jump = True; c.throttle = 1
        car2.set_controls(c); arena2.step(1)
        car2.set_controls(zero())
        for _ in range(cancel_at): arena2.step(1)
        
        # 按住 S 4 帧后松开
        for _ in range(4):
            c = zero(); c.throttle = -1
            car2.set_controls(c); arena2.step(1)
        car2.set_controls(zero())
        trace_B = []
        for t in range(60):
            arena2.step(1)
            s = snap(car2)
            trace_B.append({'t': t, 'pitch': s['pitch_deg'], 'flipping': s['is_flipping']})
        
        out[cancel_at] = {
            'hold_S_to_end': trace_A,
            'release_S_after_4f': trace_B,
        }
    return out

# ============================================================
# 主程序
# ============================================================
if __name__ == '__main__':
    print("hi")
    import sys
    which = sys.argv[1] if len(sys.argv) > 1 else 'all'
    
    if which in ('all', '1'):
        print("="*60); print("[1/4] jump hold 上限")
        r = calib_jump_hold_v2()
        json.dump(r, open('calib_jump_hold_v2.json','w'), indent=2)
        print(f"{'hold':>5} | {'max_z':>8} | {'air_frames':>10}")
        for k in sorted(r.keys()):
            print(f"{k:>5} | {r[k]['max_z']:>8.1f} | {r[k]['air_time_frames']}")
    
    if which in ('all', '2'):
        print("="*60); print("[2/4] dodge 速度加成")
        r = calib_dodge_speed_boost()
        json.dump(r, open('calib_dodge_speed_v2.json','w'), indent=2)
        print(f"{'init_vx':>8} | {'vx_after':>8} | {'gain':>8}")
        for k in sorted(r.keys()):
            print(f"{k:>8} | {r[k]['vx_after_dodge']:>8.1f} | {r[k]['gain']:>8.1f}")
    
    if which in ('all', '3'):
        print("="*60); print("[3/4] jump hold × dodge 刮地")
        r = calib_dodge_height()
        json.dump(r, open('calib_dodge_height_v2.json','w'), indent=2)
        print(f"{'hold':>5} | {'min_z':>8} | {'scraped':>8}")
        for k in sorted(r.keys()):
            print(f"{k:>5} | {r[k]['min_z_after_dodge']:>8.1f} | {str(r[k]['scraped']):>8}")
    
    if which in ('all', '4'):
        print("="*60); print("[4/4] flip cancel 逐帧曲线")
        r = calib_flip_cancel_v2()
        json.dump(r, open('calib_flip_cancel_v2.json','w'), indent=2)
        print("已保存到 calib_flip_cancel_v2.json")
        # 简略打印
        for k in sorted(r.keys()):
            A = r[k]['hold_S_to_end']
            B = r[k]['release_S_after_4f']
            pitch_A_end = A[-1]['pitch'] if A else None
            pitch_B_end = B[-1]['pitch'] if B else None
            print(f"cancel_at={k}: hold_to_end→{pitch_A_end}, release_after_4f→{pitch_B_end}")
    
    print("="*60); print("Done.")
```
